```python
# %%
import os
import cadquery as cq

def extract_cad_features(step_path):
    # STEPファイルの読み込み
    result = cq.importers.importStep(step_path)
    
    # 1. 基本幾何情報の取得
    volume = result.val().Volume()        # 体積
    area = result.val().Area()          # 表面積
    
    # 2. バウンディングボックス（外接直方体）のサイズ
    bbox = result.val().BoundingBox()
    size_x = bbox.xlen
    size_y = bbox.ylen
    size_z = bbox.zlen
    
    # 3. 意味的フィーチャの抽出（例：穴・円柱面のカウント）
    # 全ての「面(Faces)」の中から、形状が「円柱(Cylinder)」のものを抽出
    cylinders = result.faces("%cylinder").vals()
    hole_count = len(cylinders)
    
    # 4. 特徴量を辞書にまとめる
    features = {
        "file_name": os.path.basename(step_path),
        "volume": volume,
        "surface_area": area,
        "size_x": size_x,
        "size_y": size_y,
        "size_z": size_z,
        "hole_count": hole_count,
        "complexity_index": area / volume # 複雑度の指標
    }
    
    return features

# %%

```
